## 1. 실행 전에 예상하기

`my_number`를 10에서 20으로 바꾸면 무엇이 달라질까요?

- [ ] A. 그래프는 달라지지 않는다.
- [ ] B. 입력 막대만 높아진다.
- [ ] C. 입력 막대와 두 배 막대가 모두 높아진다.


```python
# ▶ 1번: Colab 실습 환경을 확인합니다. 이 셀은 수정하지 않습니다.
import sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("✅ 실습 환경 준비 완료")
print("Python:", sys.version.split()[0])
print("NumPy:", np.__version__)
print("pandas:", pd.__version__)
```

```python
# ▶ 2번: 먼저 10으로 실행한 뒤, 🟨 숫자를 20으로 바꾸어 다시 실행하세요.
my_number = 20 #@param {type:"slider", min:1, max:20, step:1}  # 🟨 여기만 바꾸세요
double_number = my_number * 2

print(f"입력한 숫자: {my_number}")
print(f"입력한 숫자의 2배: {double_number}")
```

```python
# ▶ 3번: 입력과 출력을 막대 높이로 확인합니다. 이 셀은 수정하지 않습니다.
plt.figure(figsize=(6, 3.5))
plt.bar(
    ["Input", "Input × 2"],
    [my_number, double_number],
    color=["#4C78A8", "#F58518"],
)
plt.ylim(0, 45)
plt.ylabel("Value")
plt.title("Input and Output")
plt.grid(axis="y", alpha=0.25)
plt.show()
```
